```python
# Pandas is used to read and work with our CSV dataset
import pandas as pd

# NumPy is used for calculations and handling values like N/A
import numpy as np
```

```python
# Read our new Stock Screener dataset using Pandas
df = pd.read_csv("stock_screener_final_working_dataset_400_rows.csv")

# Display the first 5 rows to confirm that the new dataset loaded correctly
df.head()
```

```python
# Check the number of rows and columns in our dataset
df.shape
```

```python
# Display all column names in our dataset
df.columns
```

```python
# Count how many missing (empty) values are present in each column
df.isnull().sum()
```

```python
# Check how many companies have zero or negative earnings
df[df["EPS_INR"] <= 0]["EPS_INR"].count()
```

```python
# Calculate P/E only when EPS is positive
# If EPS is 0 or negative, P/E is set to NaN (not applicable)
# P/E Ratio = Market Price per Share ÷ Earnings per Share (EPS)
df["PE_Ratio"] = np.where(
    df["EPS_INR"] > 0,
    df["Market_Price_INR"] / df["EPS_INR"],
    np.nan
)

# Show the first 5 P/E values
df["PE_Ratio"].head()
```

```python
# Calculate P/B Ratio = Market Price ÷ Book Value Per Share
df["PB_Ratio"] = df["Market_Price_INR"] / df["Book_Value_Per_Share_INR"]

# Display the first 5 P/B values to check our calculation
print("First 5 P/B Ratios:")
print(df["PB_Ratio"].head())
```

```python
# Calculate ROE = Net Income ÷ Shareholders' Equity × 100
# ROE = (Net Income ÷ Shareholders' Equity) × 100
df["ROE_Percent"] = (
    df["Net_Income_Crore"] / df["Shareholders_Equity_Crore"]
) * 100

# Display the first 5 ROE values to check our calculation
print("First 5 ROE Values:")
print(df["ROE_Percent"].head())
```

```python
# Calculate Debt-to-Equity = Total Debt ÷ Shareholders' Equity
df["Debt_Equity_Ratio"] = (
    df["Total_Debt_Crore"] / df["Shareholders_Equity_Crore"]
)

# Display the first 5 Debt-to-Equity values to check our calculation
print("First 5 Debt-to-Equity Ratios:")
print(df["Debt_Equity_Ratio"].head())
```

```python
# Display the names of all columns after calculating our 4 ratios
print("All columns in our dataset:")
print(df.columns.tolist())

# Display the first 5 rows of our 4 calculated ratios
print("\nFirst 5 rows of the 4 ratios:")
print(df[["PE_Ratio", "PB_Ratio", "ROE_Percent", "Debt_Equity_Ratio"]].head())
```

```python
# Save our dataset with the newly calculated 4 ratios into a new CSV file
df.to_csv("stock_screener_with_ratios.csv", index=False)

# Confirm that the file was saved successfully
print("Dataset saved successfully with all 4 ratios.")
```

```python
# Check the minimum and maximum values of our 4 calculated ratios
# This helps us identify unusually large or unusual values before using Power BI
print("P/E Range:")
print(df["PE_Ratio"].min(), "to", df["PE_Ratio"].max())

print("\nP/B Range:")
print(df["PB_Ratio"].min(), "to", df["PB_Ratio"].max())

print("\nROE Range:")
print(df["ROE_Percent"].min(), "to", df["ROE_Percent"].max())

print("\nDebt-to-Equity Range:")
print(df["Debt_Equity_Ratio"].min(), "to", df["Debt_Equity_Ratio"].max())
```

```python
# Save the complete dataset, including our calculated ratios, as a new CSV file
df.to_csv("stock_screener_final_with_ratios.csv", index=False)

# Confirm that the final file was saved successfully
print("Final dataset saved successfully!")
```
